```python
soup = BeautifulSoup(response.text)
```

```python
print(soup.prettify())
```

#### 1.	Start by using either the inspector or by viewing the page source. Can you identify a tag that might be helpful for finding the names of all performers? For now, just worry about the headliner and don't worry about the opener. (Eg. For Vince Gill, featuring Wendy Moten, we only care about Vince Gill.) Make use of this to create a list containing just the names of each inductee.

```python
perform_names = [name.get_text(strip=True) for name in soup.select("h3.title") if name.get_text(strip=True)]

print(f"Amount of performers names: {len(perform_names)}")
```

#### 2.	Next, try and find a tag that could be used to find the date and time for each show. Extract these into two lists, one containing the date and the other containing the time. (Eg. THURSDAY, AUGUST 4, 2022 AT 8:00 PM CDT should be split into August 4, 2022 and 8:00 PM CDT.)

```python
perform_hours = [hour.get_text(strip=True) for hour in soup.select("span.m-date__hour") if hour.get_text(strip=True)]

print(f"Amount of performers hours: {len(perform_hours)}")
```

```python
perform_dates = []
for block in soup.select("span.m-date__singleDate"):
    month = block.select_one("span.m-date__month")
    day   = block.select_one("span.m-date__day")
    year  = block.select_one("span.m-date__year")

    m = month.get_text(strip=True) if month else ""
    d = day.get_text(strip=True) if day else ""
    y = year.get_text(strip=True) if year else ""

    date_text = f"{m} {d}, {y}".strip()
    perform_dates.append(date_text)

print(f"Amount of performers dates: {len(perform_dates)}")
```

#### 3.	Take the two lists you created on parts 1 and 2 and convert it into a pandas DataFrame.

```python
df = pd.DataFrame({
    "Name": perform_names,
    "Date": perform_dates,
    "Time": perform_hours
})

print(df)
```
